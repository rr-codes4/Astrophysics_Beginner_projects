* **Objective:** Downloaded and inspected successfully the FITS file structure (`hdul.info()`) using Python and astropy.



```python
!pip install photutils
from astropy.utils.data import download_file
from astropy.io import fits
url = 'https://archive.stsci.edu/hlsps/jwst-ero/hlsp_jwst-ero_jwst_miri_carina_f770w_v1_i2d.fits'
local_path = download_file(url, cache=True)
hdul = fits.open(local_path)
hdul.info()
```

**What I learnt**


The output shows the real data from the JWST

*   0 PRIMARY: Has () dimensions it means it contains no pixels and only 352 character that give information about telescope's state
*   1 SCI: Holds the real infrared light measurement with dimensions (3499,1196)

*   2 ERR tracks uncertainity or noice in the pixel


*   3 CON (Context): Tells which original raw exposure were combined to make this specific pixel
*   4 WHT (Weight): Tells the data quality for each pixel

*   9 ASDF (Advanced Scientific Data Format): Advance metadata format used my JWST for coordinate modeling
